```python
###### solve for missing

def impute_education(row):
    range_map = {
        '9-12': 10,
        '13-15': 14,
        '7-8': 7,
        '15+': 15
    }
    
    education_range = row['Demographic.Years of Education']
    if pd.isna(row['Demographic.Years of Education Exact']):
        return range_map[education_range]
    else:
        return row['Demographic.Years of Education Exact']
```

```python
omics_with_demo_df_no_missing = omics_with_demo_df.copy()

omics_with_demo_df_no_missing['Demographic.Years of Education Exact'] = omics_with_demo_df_no_missing.apply(impute_education, axis=1)
omics_with_demo_df_no_missing['APOE4'] = omics_with_demo_df_no_missing['APOE4'].fillna(0)

omics_with_demo_df_no_missing = omics_with_demo_df_no_missing.drop(columns = ['Demographic.Years of Education','Demographic.ApoE genotype'])
print(omics_with_demo_df_no_missing.isna().sum().sum())
print(omics_with_demo_df_no_missing.shape)
```
