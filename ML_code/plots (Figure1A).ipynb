```python
import pandas as pd
import plotly.graph_objects as go
import plotly.io as pio
```

```python
omics_df= pd.read_csv('../preprocessed_data/omics_aibl_demo_imputed.csv',index_col=0)
```

```python
# ============================================================================
# PANEL A: Raincloud Plot using Plotly (Python)
# ============================================================================

# Load your data
# omics_df = pd.read_csv("data.csv")  # Uncomment and adjust path

# Prepare data - create Amyloid Status column
omics_df['Amyloid_Status'] = omics_df['label'].map({'H': 'High', 'N': 'Normal'})

# Define colors
COLORS = ["#E64B35", "#4DBBD5"]  # Female (red), Male (blue)
FILL_COLORS = ["rgba(230, 75, 53, 0.5)", "rgba(77, 187, 213, 0.5)"]  # With transparency

# Get unique categories
x_categories = omics_df['Amyloid_Status'].unique().tolist()
# Ensure correct order
if 'Normal' in x_categories and 'High' in x_categories:
    x_categories = ['Normal', 'High']

sub_groups = omics_df['Demographic.Sex'].unique().tolist()

# Create figure
fig = go.Figure()

# Loop through each amyloid status group
for i, group_name in enumerate(x_categories):
    group_data = omics_df[omics_df['Amyloid_Status'] == group_name]
    
    # Loop through each sex subgroup
    for j, subgroup_name in enumerate(sub_groups):
        subgroup_data = group_data[group_data['Demographic.Sex'] == subgroup_name]
        
        c_idx = j % len(COLORS)
        color = COLORS[c_idx]
        fill_color = FILL_COLORS[c_idx]
        
        # Add violin plot
        fig.add_trace(
            go.Violin(
                x=[group_name] * len(subgroup_data),
                y=subgroup_data['Centiloid'],
                name=subgroup_name,
                legendgroup=subgroup_name,
                scalegroup=subgroup_name,
                side='positive',  # Right side only
                pointpos=-0.6,    # Points on left side
                jitter=0.5,       # Spread points horizontally
                width=0.8,        # Violin width
                box_visible=False,  # No boxplot
                meanline_visible=True,  # Show mean line
                line_color=color,
                fillcolor=fill_color,
                opacity=0.8,
                marker=dict(
                    size=6,
                    line=dict(color=color, width=2),
                    color=fill_color
                ),
                offsetgroup=str(j)
            )
        )

# Add horizontal threshold line
fig.add_shape(
    type="line",
    x0=-0.5, x1=len(x_categories)-0.5,
    y0=20, y1=20,
    line=dict(color="grey", width=2, dash="dash")
)

fig.add_annotation(
    x=0,
    y=22,
    text="Threshold = 20 CL",
    showarrow=False,
    font=dict(size=12, color="grey")
)

# Update layout
fig.update_layout(
    title=dict(
        text="A. Centiloid Distribution by Amyloid Status and Sex",
        font=dict(size=16)
    ),
    width=1200,
    height=750,
    violinmode='group',
    violingroupgap=0.2,
    xaxis=dict(
        title="Amyloid Status",
        tickmode='array',
        tickvals=list(range(len(x_categories))),
        ticktext=x_categories
    ),
    yaxis=dict(
        title="Centiloid Level"
    ),
    legend=dict(
        orientation="h",
        yanchor="bottom",
        y=-0.15,
        xanchor="right",
        x=1
    ),
    plot_bgcolor='white',
    paper_bgcolor='white'
)

# Show interactive plot
fig.show()

# Save as PDF (requires kaleido: pip install kaleido)
try:
    fig.write_image("Figure1_PanelA.pdf", width=1200, height=750)
    print("✓ PDF saved successfully: Figure1_PanelA.pdf")
except Exception as e:
    print(f"PDF export failed: {e}")
    print("Installing kaleido will fix this: pip install -U kaleido")
    # Fallback to PNG
    try:
        fig.write_image("Figure1_PanelA.png", width=1200, height=750, scale=2)
        print("✓ PNG saved instead: Figure1_PanelA.png")
    except:
        print("Image export requires kaleido. Install with: pip install -U kaleido")

print("\n" + "="*60)
print("Panel A created successfully!")
print("="*60)
print("\nFiles created:")
print("1. Figure1_PanelA_Interactive.html - Open in browser")
print("2. Figure1_PanelA.pdf or .png - Static version")
print("\nLayout:")
print("- X-axis: Amyloid Status (Normal | High)")
print("- Colors: Female (red) vs Male (blue)")
print("- Points on LEFT, Distribution on RIGHT")
print("- No boxplot, Mean line visible")
```
